kita menghitung berapa banyak balok yang muncul pada setiap level tinggi, mulai dari level paling bawah hingga level tertinggi.
Perhitungan ini dilakukan karena setiap level yang dihitung akan menjadi kolom baru setelah rotasi 90°

1. Menentukan tinggi maksimum
Kita cari tahu tinggi maksimum dari seluruh kolom.
Misalkan array A = [2, 2, 1, 3, 1, 1]
Artinya:
* Kolom 1 tingginya 2
* Kolom 2 tingginya 2
* Kolom 3 tingginya 1
* Kolom 4 tingginya 3
* Kolom 5 tingginya 1
* Kolom 6 tingginya 1

2. Menghitung Jumlah balok di setiap level
Karena tinggi max adalah 3, maka ada akan terdiri dari 3 level.
Untuk setiap level kita menghitung berapa banyak kolom yang memiliki tinggi lebih dari sama dengan level tersebut.

* Level 1: Semua angka ≥ 1 → ada 6 kolom
* Level 2: Kolom yang tingginya ≥ 2 adalah: 2, 2, dan 3 → total 3 kolom
* Level 3: Kolom yang tingginya ≥ 3 hanya kolom ke-4 → total 1

Ketika nilai dijadikan dalam satu array baru setelah 1 kali rotasii.
A' = [6, 3, 1]


```python
def rotate_once(A):
    H = max(A)              # tinggi maksimum
    new_A = []              # untuk menyimpan hasil

    # hitung dari level bawah sampai atas
    for level in range(1, H + 1):
        count = 0
        for x in A:
            if x >= level:
                count += 1
        new_A.append(count)

    return new_A


def rotate_box(A, k):
    #rotasi hanya perlu sisa pembagian 4 karena rotasi 90 sebanyak 4 kali akan mengambalikan posisi semula
    #menggunakan % atau mod sebagai sisa hasil pembagian
    #hasil sisa bagi ini digunakan sebagai berapa kali harus melakukan rotate_once
    #Misalkan 5 dibagi 4 adalah 1 sisa 1. Kalau sisa hasil 0 (4%4 = 0) maka balik ke posisi awal atau tidak di rotasi

    k = k % 4
    for _ in range(k):
        A = rotate_once(A)
    return A


# =============== MAIN INPUT ===============
N = int(input("Masukkan jumlah baris:")) #jumlah baris
A = list(map(int, input("Masukkan isi array (dipisahkan dengan spasi):").split())) #isi array A sebanyak N angka atau sebanyak jumlah barisnya
k = int(input("Masukkan jumlah rotasi:")) #jumlah rotasi

result = rotate_box(A, k)
print(*result)
```
